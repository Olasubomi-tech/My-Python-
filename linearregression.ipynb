```python
import numpy as np
import math 
def bisection(f, a, b, tol, N):
    if f(a) * f(b) >= 0:
        print("Bisection fails: f(a) and f(b) must have opposite signs.")
        return
    n = 1
    p = 0
    while n <= N:
        p = a + (b - a) / 2
        if np.isclose(f(p), 0) or np.abs(b - a) / 2 < tol:
            print("We have that our root is", p, "and the number of iterations is", n)
            return
        if f(a) * f(p) < 0:
            b = p
        else:
            a = p
        n += 1
    print("Failed to converge. Best guess for root is", p, "after", N, "iterations")
bisection(lambda x: x**3+2*x**3-5*x-2, -3, 0, 1e-4, 200)
```

```python
import numpy as np
def fixedpt(g, pzero, eps, N):
    n = 1
    while n<N:
        print(n)
        pone = g(pzero)
        if np.abs(pone-pzero)<eps:
            print('p is ', pone, ' and the iteration number is ', n)
            return
        pzero = pone
        n += 1
    print('Did not converge. The last estimate is p = ', pzero)
```

Lagrange Interpolation 

```python
import numpy as np
from numpy import pi
import matplotlib.pyplot as plt


def f_1(x):
    return np.sin(np.abs(x))

def f_2(x):
    return np.sign(x) * np.cos(x)

def chebyshev_nodes(n):
    j = np.arange(n)
    return np.cos((2 * j + 1) * np.pi / (2 * n))

def lagrange_interpolation(x_nodes, y_nodes, x_eval):
    n = len(x_nodes)
    j = np.arange(n)
    weights = (-1)**j * np.sin((2 * j + 1) * np.pi / (2 * n))

    x_eval = np.asarray(x_eval)
    P_eval = np.zeros_like(x_eval, dtype=float)

    for i, x in enumerate(x_eval):
        exact = np.isclose(x - x_nodes, 0, atol=1e-15)
        if np.any(exact):
            P_eval[i] = y_nodes[exact][0]
        else:
            P_eval[i] = np.sum((weights / (x - x_nodes)) * y_nodes) / np.sum(weights / (x - x_nodes))
            
    return P_eval
a_k = np.linspace(-1, 1, 1000)

n_values = np.arange(1, 101)
error_inf_function1 = []
error_inf_function2 = []

for n in n_values:
    nodes = chebyshev_nodes(n)
    y1_nodes = f_1(nodes)
    P1 = lagrange_interpolation(nodes, y1_nodes, a_k)
    error_inf_function1.append(np.max(np.abs(f_1(a_k) - P1)))
    y2_nodes = f_2(nodes)
    P2 = lagrange_interpolation(nodes, y2_nodes, a_k)
    error_inf_function2.append(np.max(np.abs(f_2(a_k) - P2)))

print(f"The L^infinity error for the first function is {error_inf_function1[-1]}")
print(f"The L^infinity error for the second function is {error_inf_function2[-1]}")

n_demo = 21
nodes_demo = chebyshev_nodes(n_demo)

fig1, ax1 = plt.subplots(figsize=(8, 5))
ax1.plot(a_k, f_1(a_k), 'b--', linewidth=1.5, label=r'(i) $f(x) = \sin(|x|)$')
ax1.plot(a_k, lagrange_interpolation(nodes_demo, f_1(nodes_demo), a_k), 'b-', label=r'(i) $P_{19}(x)$')
ax1.plot(a_k, f_2(a_k), 'r--', linewidth=1.5, label=r'(ii) $f(x) = \mathrm{sign}(x)\cos(x)$')
ax1.plot(a_k, lagrange_interpolation(nodes_demo, f_2(nodes_demo), a_k), 'r-', label=r'(ii) $P_{19}(x)$')
ax1.set_title(f"Polynomial Interpolation at Chebyshev Points ($n = {n_demo}$)")
ax1.set_xlabel("$x$")
ax1.set_ylabel("$y$")
ax1.set_xlim(-1, 1)
ax1.set_ylim(-2, 2)
ax1.grid(True)
ax1.legend()
plt.tight_layout()
plt.show()

fig2, ax2 = plt.subplots(figsize=(8, 5))
ax2.semilogy(n_values, error_inf_function1, 'b-o', markersize=3, label=r'(i) $f(x) = \sin(|x|)$')
ax2.semilogy(n_values, error_inf_function2, 'r-s', markersize=3, label=r'(ii) $f(x) = \mathrm{sign}(x)\cos(x)$')
ax2.set_title(r"Interpolation Error $\|f - P_{n-1}\|_\infty$ vs $n$ (Log Scale)")
ax2.set_xlabel("Number of nodes $n$")
ax2.set_ylabel(r"$\|f - P_{n-1}\|_\infty$")
ax2.set_xlim(1, 30)
ax2.set_ylim(1e-1, 1e4)
ax2.grid(True)
ax2.legend()

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import linregress

# =============================================================================
# Table 1: Legendre-Gauss Nodes (x_i) and Weights (w_i) for n = 0 to 7
# =============================================================================
table_data = {
    0: (np.array([0.0]), np.array([2.0])),
    1: (np.array([-0.5773502691896257, 0.5773502691896257]), 
        np.array([1.0, 1.0])),
    2: (np.array([-0.7745966692414834, 0.0, 0.7745966692414834]), 
        np.array([0.5555555555555556, 0.8888888888888888, 0.5555555555555556])),
    3: (np.array([-0.8611363115940526, -0.3399810435848563, 0.3399810435848563, 0.8611363115940526]), 
        np.array([0.3478548451374538, 0.6521451548625461, 0.6521451548625461, 0.3478548451374538])),
    4: (np.array([-0.9061798459386640, -0.5384693101056831, 0.0, 0.5384693101056831, 0.9061798459386640]),
        np.array([0.2369268850561891, 0.4786286704993665, 0.5688888888888889, 0.4786286704993665, 0.2369268850561891])),
    5: (np.array([-0.9324695142031521, -0.6612093864662645, -0.2386191860831969, 0.2386191860831969, 0.6612093864662645, 0.9324695142031521]),
        np.array([0.1713244923791704, 0.3607615730481386, 0.4679139345726910, 0.4679139345726910, 0.3607615730481386, 0.1713244923791704])),
    6: (np.array([-0.9491079123427585, -0.7415311855993945, -0.4058451513773972, 0.0, 0.4058451513773972, 0.7415311855993945, 0.9491079123427585]),
        np.array([0.1294849661688697, 0.2797053914892766, 0.3818300505051189, 0.4179591836734694, 0.3818300505051189, 0.2797053914892766, 0.1294849661688697])),
    7: (np.array([-0.9602898564975363, -0.7966664774136267, -0.5255324099163290, -0.1834346424956498, 0.1834346424956498, 0.5255324099163290, 0.7966664774136267, 0.9602898564975363]),
        np.array([0.1012285362903763, 0.2223810344533745, 0.3137066458778873, 0.3626837833783620, 0.3626837833783620, 0.3137066458778873, 0.2223810344533745, 0.1012285362903763]))
}

# =============================================================================
# Problem 1: Gaussian Quadrature for int_0^4 exp(x) dx
# =============================================================================
def problem_1():
    print("--- PROBLEM 1 ---")
    f = np.exp
    a, b = 0.0, 4.0
    I_exact = np.exp(4.0) - 1.0
    
    n_vals = list(range(8))
    errors = []
    
    for n in n_vals:
        r, w = table_data[n]
        x = (b - a) / 2.0 * r + (a + b) / 2.0
        I_num = (b - a) / 2.0 * np.sum(w * f(x))
        err = abs(I_num - I_exact)
        errors.append(err)
        print(f"n = {n}: Approx = {I_num:.10f}, Error = {err:.6e}")
        
    plt.figure(figsize=(7, 4.5))
    plt.semilogy(n_vals, errors, 'o-', color='navy', linewidth=2)
    plt.xlabel('Degree index \(n\)')
    plt.ylabel('Absolute Error')
    plt.title('Problem 1: Error vs. \(n\) for \(\int_0^4 e^x dx\)')
    plt.grid(True, which="both", ls="--")
    plt.show()

# =============================================================================
# Problem 2: Composite Gaussian Quadrature
# =============================================================================
def composite_gauss(f, a, b, m, n):
    h = (b - a) / m
    r, w = table_data[n]
    total = 0.0
    for j in range(m):
        xj_prev = a + j * h
        xj = a + (j + 1) * h
        nodes = (h / 2.0) * r + (xj_prev + xj) / 2.0
        total += (h / 2.0) * np.sum(w * f(nodes))
    return total

def problem_2():
    print("\n--- PROBLEM 2 ---")
    m_list = [2**k for k in range(1, 11)]
    
    # Part A: Smooth Function
    f1 = lambda x: 101.0 * np.exp(x) * np.sin(10.0 * x)
    I1_exact = np.exp(4.0) * (np.sin(40.0) - 10.0 * np.cos(40.0)) + 10.0
    h1_list = [4.0 / m for m in m_list]
    
    plt.figure(figsize=(7, 4.5))
    print("Part 1: Linear Regression Fits for 101*exp(x)*sin(10x)")
    for n in range(4): # Plotting for n = 0, 1, 2, 3
        rel_errors = []
        for m in m_list:
            val = composite_gauss(f1, 0.0, 4.0, m, n)
            rel_err = abs(val - I1_exact) / abs(I1_exact)
            rel_errors.append(rel_err)
        
        # Fit linear regression log(y) = log(alpha) + beta * log(h)
        slope, intercept, _, _, _ = linregress(np.log(h1_list), np.log(rel_errors))
        alpha = np.exp(intercept)
        print(f"n = {n}: beta = {slope:.4f}, alpha = {alpha:.4e} (Theoretical beta = {2*n+2})")
        plt.loglog(h1_list, rel_errors, 'o-', label=f'n={n} (fitted \(\\beta\)={slope:.2f})')
        
    plt.xlabel('Step size \(h\)')
    plt.ylabel('Relative Error')
    plt.title('Problem 2A: Relative Error vs. \(h\) (Smooth Function)')
    plt.legend()
    plt.grid(True, which="both", ls="--")
    plt.show()

    # Part B: Discontinuous Function sign(x)
    f2 = lambda x: np.sign(x)
    I2_exact = 0.5
    h2_list = [1.5 / m for m in m_list]
    
    plt.figure(figsize=(7, 4.5))
    print("\nPart 2: Linear Regression Fits for sign(x)")
    for n in range(4):
        rel_errors = []
        for m in m_list:
            val = composite_gauss(f2, -0.5, 1.0, m, n)
            rel_err = abs(val - I2_exact) / I2_exact
            rel_errors.append(rel_err)
            
        slope, intercept, _, _, _ = linregress(np.log(h2_list), np.log(rel_errors))
        print(f"n = {n}: beta = {slope:.4f}, alpha = {np.exp(intercept):.4e}")
        plt.loglog(h2_list, rel_errors, 's-', label=f'n={n} (fitted \(\\beta\)={slope:.2f})')
        
    plt.xlabel('Step size \(h\)')
    plt.ylabel('Relative Error')
    plt.title('Problem 2B: Relative Error vs. \(h\) for sign(x)')
    plt.legend()
    plt.grid(True, which="both", ls="--")
    plt.show()

# =============================================================================
# Problems 5 & 6: ODE Solvers (Backward Euler & Crank-Nicolson)
# =============================================================================
def exact_y(t):
    return 2.0 / (5.0 - 4.0 * np.sin(np.pi * t / 2.0))

def solve_be(n_steps, T=5.0, y0=0.4):
    h = T / n_steps
    t = np.linspace(0, T, n_steps + 1)
    y = np.zeros(n_steps + 1)
    y[0] = y0
    
    for k in range(n_steps):
        tk1 = t[k+1]
        A = h * np.pi * np.cos(np.pi * tk1 / 2.0)
        yk = y[k]
        if abs(A) < 1e-12:
            y[k+1] = yk
        else:
            disc = 1.0 - 4.0 * A * yk
            y[k+1] = (1.0 - np.sqrt(disc)) / (2.0 * A)
    return t, y

def solve_cn(n_steps, T=5.0, y0=0.4):
    h = T / n_steps
    t = np.linspace(0, T, n_steps + 1)
    y = np.zeros(n_steps + 1)
    y[0] = y0
    
    for k in range(n_steps):
        tk = t[k]
        tk1 = t[k+1]
        yk = y[k]
        f_k = np.pi * (yk**2) * np.cos(np.pi * tk / 2.0)
        A = 0.5 * h * np.pi * np.cos(np.pi * tk1 / 2.0)
        C = yk + 0.5 * h * f_k
        if abs(A) < 1e-12:
            y[k+1] = C
        else:
            disc = 1.0 - 4.0 * A * C
            y[k+1] = (1.0 - np.sqrt(disc)) / (2.0 * A)
    return t, y

def problem_5():
    print("\n--- PROBLEM 5 ---")
    n = 1000
    t_be, y_be = solve_be(n)
    t_cn, y_cn = solve_cn(n)
    t_exact = np.linspace(0, 5.0, 1000)
    
    plt.figure(figsize=(8, 5))
    plt.plot(t_exact, exact_y(t_exact), 'k-', linewidth=2, label='Exact Solution')
    plt.plot(t_be, y_be, 'r--', label='Backward Euler (n=1000)')
    plt.plot(t_cn, y_cn, 'b:', label='Crank-Nicolson (n=1000)')
    plt.xlabel('Time \(t\)')
    plt.ylabel('\(y(t)\)')
    plt.title('Problem 5: Comparison of Numerical Methods (n = 1000)')
    plt.legend()
    plt.grid(True)
    plt.show()

def problem_6():
    print("\n--- PROBLEM 6 ---")
    m_powers = range(1, 12)  # m = 2^1 to 2^11
    h_vals = []
    err_be = []
    err_cn = []
    
    y_true_T5 = exact_y(5.0)
    
    for m in m_powers:
        m_val = 2**m
        h = 1.0 / (10.0 * m_val)
        n_steps = int(round(5.0 / h))
        
        _, y_b = solve_be(n_steps)
        _, y_c = solve_cn(n_steps)
        
        h_vals.append(h)
        err_be.append(abs(y_b[-1] - y_true_T5))
        err_cn.append(abs(y_c[-1] - y_true_T5))
        
    # Fits
    slope_be, intercept_be, _, _, _ = linregress(np.log(h_vals), np.log(err_be))
    slope_cn, intercept_cn, _, _, _ = linregress(np.log(h_vals), np.log(err_cn))
    
    print(f"Backward Euler: beta = {slope_be:.4f}, alpha = {np.exp(intercept_be):.4e}")
    print(f"Crank-Nicolson: beta = {slope_cn:.4f}, alpha = {np.exp(intercept_cn):.4e}")
    
    plt.figure(figsize=(7, 4.5))
    plt.loglog(h_vals, err_be, 'ro-', label=f'Backward Euler (fitted \(\\beta\)={slope_be:.2f})')
    plt.loglog(h_vals, err_cn, 'bs-', label=f'Crank-Nicolson (fitted \(\\beta\)={slope_cn:.2f})')
    plt.xlabel('Step Size \(h\)')
    plt.ylabel('Absolute Error at \(T=5\)')
    plt.title('Problem 6: Convergence Rate Analysis for ODE Solvers')
    plt.legend()
    plt.grid(True, which="both", ls="--")
    plt.show()

# =============================================================================
# Main Execution
# =============================================================================
if __name__ == "__main__":
    problem_1()
    problem_2()
    problem_5()
    problem_6()
```
